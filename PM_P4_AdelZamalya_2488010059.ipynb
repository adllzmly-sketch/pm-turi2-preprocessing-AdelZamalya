# Dataset

```python
import pandas as pd

data = {
    'usia': [23, 25, None, 30, 28, 35, None, 27, 32, 24],
    'pendapatan': [3000000, 4000000, 3500000, None, 5000000, 6000000, 4500000, None, 7000000, 3200000],
    'pendidikan': ['SMA', 'S1', 'S1', 'S2', 'S1', 'S2', 'SMA', 'S1', 'S2', 'SMA'],
    'kota': ['Cirebon', 'Kuningan', 'Majalengka', 'Cirebon', 'Kuningan',
             'Majalengka', 'Cirebon', 'Kuningan', 'Majalengka', 'Cirebon'],
    'membeli': ['Tidak', 'Ya', 'Ya', 'Ya', 'Tidak', 'Ya', 'Tidak', 'Ya', 'Ya', 'Tidak']
}

df = pd.DataFrame(data)

df
```

# Langkah 1

```python
df['usia'] = df['usia'].fillna(df['usia'].median())
df['pendapatan'] = df['pendapatan'].fillna(df['pendapatan'].median())

display(df)
```

# Langkah 2

```python
x = df.drop(columns=['membeli'])
y = df['membeli']

print("Fitur (X):")
display(x)

print("Label (y):")
display(y)
```

# Langkah 3

```python
x['pendidikan'] = x['pendidikan'].map({'SMA':0, 'S1' :1, 'S2':2})
x = pd.get_dummies(x, columns=['kota'], dtype=int)
y = y.map({'Tidak':0, 'Ya':1})

print("X setelah encoding:")
display(x)

print("y setelah encoding:")
display(y)
```

# Langkah 4

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(
    x, y, test_size=0.3, random_state=42
)

print("Pembagian Data:")
print("Jumlah data X_train:", len(x_train))
print("Jumlah data X_test:", len(x_test))
print("Jumlah data y_train:", len(y_train))
print("Jumlah data y_test:", len(y_test))
```

# Langkah 5

```python
from sklearn.preprocessing import StandardScaler
num = ['usia', 'pendapatan', 'pendidikan']
sc = StandardScaler()
x_train[num] = sc.fit_transform(x_train[num])
x_test[num] = sc.transform(x_test[num])

print("Data setelah scaling:")
display(x_train)
display(x_test)
```

# Langkah 6

```python
print("Jumlah nilai hilang:")
print(x_train.isnull().sum())

print("\nTipe data:")
print(x_train.dtypes)

print("\nStatistik data setelah scaling:")
display(x_train.describe())
```
